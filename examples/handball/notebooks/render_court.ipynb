```python
import sys
from pathlib import Path

current_dir = Path.cwd().resolve()
repo_root = next(
    parent for parent in [current_dir, *current_dir.parents]
    if (parent / "sports").is_dir()
)
sys.path.insert(0, str(repo_root))
```

```python
import cv2
import numpy as np
import supervision as sv

from sports.annotators.handball import draw_court, draw_paths_on_court, draw_points_on_court
from sports.configs.handball import HandballCourtConfiguration
```

```python
config = HandballCourtConfiguration()
court = draw_court(config=config, padding=80, line_thickness=6, scale=0.2)
court.shape
```

```python
xy = np.array([
    [650, 750],
    [1050, 1220],
    [2100, 980],
    [3300, 840],
])

court = draw_points_on_court(
    config=config,
    xy=xy,
    face_color=sv.Color.from_hex("#E53935"),
    edge_color=sv.Color.WHITE,
    radius=12,
    thickness=3,
    padding=80,
    scale=0.2,
    court=court,
)

paths = [np.array([[650, 750], [950, 840], [1280, 760], [1600, 900], [2100, 980]])]
court = draw_paths_on_court(
    config=config,
    paths=paths,
    color=sv.Color.WHITE,
    thickness=4,
    padding=80,
    scale=0.2,
    court=court,
)
```

```python
cv2.imwrite("handball-court.png", court)
```
