```python
import os
from pathlib import Path

from roboflow import Roboflow

EXAMPLE_DIR = Path.cwd().resolve().parent
DATASET_DIR = EXAMPLE_DIR / "data" / "datasets" / "handball-player-detection"
RUNS_DIR = EXAMPLE_DIR / "data" / "runs"

ROBOFLOW_API_KEY = os.environ.get("ROBOFLOW_API_KEY", "YOUR_API_KEY")
rf = Roboflow(api_key=ROBOFLOW_API_KEY)

project = rf.workspace("project-k1mun").project("handball_player_detection")
dataset = project.version(2).download("yolov8", location=str(DATASET_DIR))
dataset.location
```

```python
!yolo task=detect mode=train model=yolov8n.pt data={dataset.location}/data.yaml epochs=100 imgsz=640 project={RUNS_DIR} name=handball-player-detection
```
