```python
import os
from pathlib import Path

from roboflow import Roboflow

EXAMPLE_DIR = Path.cwd().resolve().parent
DATASET_DIR = EXAMPLE_DIR / "data" / "datasets" / "handball-court-keypoints"
RUNS_DIR = EXAMPLE_DIR / "data" / "runs"

ROBOFLOW_API_KEY = os.environ.get("ROBOFLOW_API_KEY", "YOUR_API_KEY")
rf = Roboflow(api_key=ROBOFLOW_API_KEY)

project = rf.workspace("handball-wckh8").project("handball-keypoint-detection-ndvsl")
dataset = project.version(7).download("yolov8", location=str(DATASET_DIR))
dataset.location
```

```python
!yolo task=pose mode=train model=yolov8n-pose.pt data={dataset.location}/data.yaml epochs=100 imgsz=640 project={RUNS_DIR} name=handball-court-keypoints
```
