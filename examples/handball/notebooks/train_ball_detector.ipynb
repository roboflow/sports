This notebook downloads one of the Roboflow Universe handball ball datasets and trains a YOLO detector.

```python
import os
from pathlib import Path

from roboflow import Roboflow

BALL_DATASETS = {
    "handball_ballon": {
        "workspace": "handballdetection",
        "project": "handball-ballon",
        "version": 7,
        "directory": "handball-ballon",
    },
    "ball_only_detection": {
        "workspace": "handball-detection",
        "project": "ball-only-detection",
        "version": 1,
        "directory": "ball-only-detection",
    },
}

SELECTED_DATASET = "handball_ballon"
EXAMPLE_DIR = Path.cwd().resolve().parent
RUNS_DIR = EXAMPLE_DIR / "data" / "runs"
ROBOFLOW_API_KEY = os.environ.get("ROBOFLOW_API_KEY", "YOUR_API_KEY")

dataset_config = BALL_DATASETS[SELECTED_DATASET]
DATASET_DIR = EXAMPLE_DIR / "data" / "datasets" / dataset_config["directory"]
rf = Roboflow(api_key=ROBOFLOW_API_KEY)
project = rf.workspace(dataset_config["workspace"]).project(dataset_config["project"])
dataset = project.version(dataset_config["version"]).download("yolov8", location=str(DATASET_DIR))
dataset.location
```

```python
!yolo task=detect mode=train model=yolov8n.pt data={dataset.location}/data.yaml epochs=100 imgsz=640 project={RUNS_DIR} name=handball-ball-detection
```
